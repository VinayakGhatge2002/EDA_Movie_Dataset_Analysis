```python
# Relationshops
sns.scatterplot(data = df , x = "BudgetUSD", y = "Global_BoxOfficeUSD", alpha = 0.4)
plt.title("Budget UDS vs Global Box Office")
plt.xscale("log")
plt.yscale("log")
plt.show()
```

```python
sns.scatterplot(data = df , x = "IMDbRating", y = "RottenTomatoesScore", alpha = 0.3)
plt.title("IMb Rating vs Rotten Tomatoes Score")
plt.show()
```

```python
# Groupby analyses
genre_year_stats = df.groupby(["Genre", "ReleaseYear"]).agg({
    "Global_BoxOfficeUSD": "mean",
    "BudgetUSD": "mean",
    "IMDbRating": "mean",
    "RottenTomatoesScore": "mean"
}).reset_index()
```

```python
genre_year_stats
```

```python
df.columns
```

```python
# line plot for genere by release data
plt.figure(figsize = (15,10))
sns.lineplot(data = genre_year_stats, x = "ReleaseYear", y = "Global_BoxOfficeUSD", hue = 'Genre')
plt.title("Avgerag General Box office by Genre Over Year")
plt.show()
```

```python
# get specific gener line graph
plt.figure(figsize = (15,10))
sns.lineplot(data = genre_year_stats.query("Genre == 'Action'") , x = "ReleaseYear", y = "Global_BoxOfficeUSD", hue = 'Genre')
plt.title("Avgerag General Box office by Genre Over Year")
plt.show()
```

```python
plt.figure(figsize = (15,10))
sns.lineplot(data = genre_year_stats.query("Genre == 'Sci-Fi'") , x = "ReleaseYear", y = "Global_BoxOfficeUSD", hue = 'Genre')
plt.title("Avgerag General Box office by Genre Over Year")
plt.show()
```

```python
plt.figure(figsize = (15,10))
sns.lineplot(data = genre_year_stats.query("Genre == 'Horror'") , x = "ReleaseYear", y = "Global_BoxOfficeUSD", hue = 'Genre')
plt.title("Avgerag General Box office by Genre Over Year")
plt.show()
```
